# 2. Basic Structure of a Matplotlib Plot

The main concepts are:

- **Figure** → complete canvas/window
- **Axes** → area where the plot is drawn
- **Axis** → x-axis and y-axis scales
- **Artists** → visual elements such as lines, text, markers, and legends

For beginners, remember:

**Figure = canvas**

**Axes = plotting area**

**Axis = x/y scale**


# 3. Basic Plotting with Pyplot

## Line Plot

A line plot is useful for showing trends or changes over an ordered sequence.

Use `plt.plot()`.


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
````

## Labels, Title, Legend, and Grid

Important functions:
- `plt.xlabel()` → x-axis label
- `plt.ylabel()` → y-axis label
- `plt.title()` → chart title
- `plt.legend()` → identifies plotted data
- `plt.grid()` → displays grid lines


````python
plt.plot(x, y, label="Sales")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.legend()
plt.grid(True)

plt.show()
````

## Formatting a Line

Common options:
- `color` → line color
- `marker` → marker shape
- `linestyle` → line style


````python
plt.plot(
    x, y,
    color="blue",
    marker="o",
    linestyle="--",
    label="Sales"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Formatted Line Plot")
plt.legend()
plt.grid(True)
plt.show()
````

## Axis Limits

Use:
- `plt.xlim()` → x-axis limits
- `plt.ylim()` → y-axis limits


````python
plt.plot(x, y)

plt.xlim(1, 5)
plt.ylim(0, 35)

plt.title("Axis Limits")
plt.show()
````

# 4. Figure Size and DPI

- `figsize=(width, height)` controls figure size in inches.
- `dpi` controls resolution.

Higher DPI is useful when saving high-quality figures.


````python
plt.figure(figsize=(8, 5), dpi=100)

plt.plot(x, y)
plt.title("Figure Size and DPI")
plt.show()
````

# 5. Multiple Subplots

`plt.subplots()` is the recommended approach for creating multiple plots.

It returns:
- `fig` → Figure
- `ax` → Axes


````python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(x, y)
ax[0].set_title("Line Plot")

ax[1].bar(x, y)
ax[1].set_title("Bar Plot")

plt.tight_layout()
plt.show()
````

## `subplot()`

`plt.subplot()` is another way to create multiple plotting areas. You may see it in existing code.


````python
plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Line")

plt.subplot(1, 2, 2)
plt.bar(x, y)
plt.title("Bar")

plt.tight_layout()
plt.show()
````

# 6. Different Types of Plots

| Plot | Main Purpose |
|---|---|
| Line | Show trends |
| Bar | Compare categories |
| Histogram | Show numerical distribution |
| Scatter | Show relationship between two variables |
| Pie | Show simple proportions |
| Box Plot | Show distribution and potential outliers |
| Heatmap-style image | Show patterns in a matrix |


## 6.1 Line Plot

Use `plt.plot()` to show trends over an ordered sequence.


````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 45, 50]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.grid(True)
plt.show()
````

## 6.2 Bar Plot

Use:
- `plt.bar()` → vertical bars
- `plt.barh()` → horizontal bars

Bar plots are useful for comparing categories.


````python
categories = ["IT", "HR", "Finance", "Sales"]
employees = [40, 25, 20, 35]

plt.bar(categories, employees)
plt.xlabel("Department")
plt.ylabel("Employees")
plt.title("Employees by Department")
plt.show()
````

````python
plt.barh(categories, employees)
plt.xlabel("Employees")
plt.ylabel("Department")
plt.title("Employees by Department")
plt.show()
````

## 6.3 Histogram

A histogram shows the **distribution** of numerical data by grouping values into intervals called bins.

Use `plt.hist()`.


````python
ages = [22, 24, 25, 25, 26, 27, 28, 30, 31, 32, 35, 36]

plt.hist(ages, bins=5)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
````

## 6.4 Scatter Plot

A scatter plot shows the relationship between two numerical variables.

Use `plt.scatter()`.


````python
study_hours = [1, 2, 3, 4, 5, 6]
scores = [45, 50, 55, 65, 70, 80]

plt.scatter(study_hours, scores)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
````

## 6.5 Pie Chart

A pie chart shows parts of a whole.

Use it mainly when there are only a few categories and the proportions are easy to compare.


````python
labels = ["IT", "HR", "Finance", "Sales"]
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()
````

## 6.6 Box Plot

A box plot summarizes a numerical distribution and helps identify potential outliers.

Use `plt.boxplot()`.

Important parts include the median, quartiles, whiskers, and potential outliers.


````python
salaries = [25000, 28000, 30000, 32000, 35000, 36000, 38000, 40000, 90000]

plt.boxplot(salaries)
plt.ylabel("Salary")
plt.title("Salary Distribution")
plt.show()
````

## 6.7 Heatmap-Style Visualization with `imshow()`

Matplotlib can display a matrix using `plt.imshow()`.

This is useful for visualizing values arranged in rows and columns.


````python
data_matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

plt.imshow(data_matrix)
plt.colorbar()
plt.title("Matrix Visualization")
plt.show()
````

# 7. Ticks, Tick Labels, Text, and Annotations

Useful functions:
- `plt.xticks()` → customize x-axis ticks
- `plt.yticks()` → customize y-axis ticks
- `plt.text()` → place text at a position
- `plt.annotate()` → add an explanation pointing to a location


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker="o")
plt.xticks([1, 2, 3, 4, 5])
plt.yticks([0, 10, 20, 30, 40])
plt.text(4, 30, "Highest Point")

plt.title("Custom Ticks and Text")
plt.grid(True)
plt.show()
````

````python
plt.plot(x, y, marker="o")

plt.annotate(
    "Peak",
    xy=(4, 30),
    xytext=(3, 35),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Annotation Example")
plt.show()
````

# 8. Figure and Axes Objects

For more control, use the object-oriented style:

```python
fig, ax = plt.subplots()
```

Then use methods such as:
- `ax.plot()`
- `ax.set_title()`
- `ax.set_xlabel()`
- `ax.set_ylabel()`

This is especially useful for complex figures and multiple subplots.


````python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker="o")
ax.set_title("Object-Oriented Matplotlib")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.grid(True)

plt.show()
````

# 9. Dates and Time Series

Matplotlib can visualize data recorded over time.

Typical steps:
1. Convert dates to datetime values
2. Plot the values
3. Format the date labels when needed

`matplotlib.dates` provides tools for date formatting.


````python
import matplotlib.pyplot as plt
import pandas as pd

dates = pd.date_range("2026-01-01", periods=5)
sales = [10, 15, 12, 20, 18]

plt.plot(dates, sales, marker="o")

plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
````

# 10. Matplotlib with Pandas

Pandas DataFrames have a built-in `.plot()` method that uses Matplotlib underneath.

This is convenient when your data is already stored in a DataFrame.


````python
import pandas as pd

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [20, 35, 30, 45]
})

df.plot(x="Month", y="Sales", kind="line", marker="o")
plt.title("Sales Using Pandas")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
````

### Matplotlib
- More control over individual plot elements
- General-purpose visualization library
- Excellent for detailed customization

### Seaborn
- Built on top of Matplotlib
- Provides higher-level statistical plots
- Often requires less code for statistical visualizations

# 11. Global Customization with rcParams

`rcParams` contains Matplotlib's global configuration settings.

It can change defaults such as:
- Figure size
- Font size
- Line width

Global settings affect plots created later in the current session, so use them carefully.


````python
plt.rcParams["figure.figsize"] = (7, 4)
plt.rcParams["font.size"] = 11

plt.plot(x, y, marker="o")
plt.title("Using rcParams")
plt.show()
````

# 12. Saving Figures

Use:

`plt.savefig()`

Common formats:
- PNG
- PDF
- SVG
- JPG/JPEG

Use `dpi` for resolution.

Example:

`plt.savefig("plot.png", dpi=300, bbox_inches="tight")`


````python
plt.figure(figsize=(7, 4))
plt.plot(x, y, marker="o")
plt.title("Figure to Save")

plt.savefig("my_plot.png", dpi=300)
plt.show()
````
