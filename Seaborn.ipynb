# Seaborn

Simple, practical notes and examples for learning Seaborn from scratch.

## 1. Import Libraries

### 📝 What is this cell doing?
We import Pandas, Seaborn, and Matplotlib and set a clean Seaborn theme.

**Why do we need it?**
- Pandas → work with tabular data.
- Seaborn → create statistical visualizations.
- Matplotlib → customize and display plots.

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
````

## 2. Create a Small Dataset

### 📝 What is this cell doing?
We create a small DataFrame containing months, sales, study hours, marks, and departments.

**Why do we need it?**
This gives us simple data to practice different Seaborn plots without needing a large real-world dataset.

````python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [100, 120, 150, 140, 170, 200],
    "hours": [2, 3, 4, 5, 6, 7],
    "marks": [45, 50, 58, 65, 72, 85],
    "department": ["IT", "HR", "IT", "Sales", "HR", "IT"]
})

df
````

## 3. Basic Seaborn Plots

### Line Plot — trends

### Line Plot
**What is it?** A line plot connects data points in order to show how a value changes.

**What does this plot show?** Monthly sales from January to June.

**How does it help us?** It makes trends easy to see — for example, whether sales are increasing, decreasing, or fluctuating over time.

**Remember:** Use `lineplot()` mainly when the order of the x-axis matters, especially for time or progression.

````python
sns.lineplot(data=df, x="month", y="sales")
plt.title("Monthly Sales")
plt.show()
````

### Bar Plot — compare categories

### Bar Plot
**What is it?** A bar plot compares a numerical value across different categories.

**What does this plot show?** Average sales for each department.

**How does it help us?** It lets us quickly compare categories and see which department has higher or lower values.

**Remember:** Use `barplot()` when your main goal is category comparison.

````python
sns.barplot(data=df, x="department", y="sales")
plt.title("Sales by Department")
plt.show()
````

### Histogram — distribution

### Histogram
**What is it?** A histogram shows how numerical values are distributed by grouping them into ranges called bins.

**What does this plot show?** How the students' marks are distributed.

**How does it help us?** It helps us understand where most values are concentrated and whether the data is spread out or clustered.

**Remember:** Use `histplot()` when you want to understand the distribution of one numerical variable.

````python
sns.histplot(data=df, x="marks", bins=5)
plt.title("Marks Distribution")
plt.show()
````

### KDE Plot — smooth distribution

### KDE Plot
**What is it?** A KDE (Kernel Density Estimate) plot shows a smooth estimate of the distribution of numerical data.

**What does this plot show?** The density of the students' marks.

**How does it help us?** It gives us a smooth view of where values are more or less concentrated compared with the bars of a histogram.

**Remember:** Use `kdeplot()` when you want a smooth representation of a distribution.

````python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [100, 120, 150, 140, 170, 200],
    "hours": [2, 3, 4, 5, 6, 7],
    "marks": [25, 20, 28, 25, 42, 85],
    "department": ["IT", "HR", "IT", "Sales", "HR", "IT"]
})

df
````

````python
sns.kdeplot(data=df, x="marks")
plt.title("Marks Density")
plt.show()
````

### Scatter Plot — relationship between two variables

### Scatter Plot
**What is it?** A scatter plot places one observation as a point using an x-value and a y-value.

**What does this plot show?** The relationship between study hours and marks.

**How does it help us?** It helps us see whether two numerical variables have a relationship, such as whether marks tend to increase as study hours increase.

**Remember:** Use `scatterplot()` to investigate relationships between two numerical variables.

````python
sns.scatterplot(data=df, x="hours", y="marks")
plt.title("Study Hours vs Marks")
plt.show()
````

### Box Plot — spread and possible outliers

### Box Plot
**What is it?** A box plot summarizes the spread of numerical data using the median, quartiles, and possible outliers.

**What does this plot show?** The spread of students' marks.

**How does it help us?** It quickly shows the center of the data, how spread out it is, and whether there may be unusual values.

**Remember:** Use `boxplot()` when you want to compare distributions or look for possible outliers.

````python
sns.boxplot(data=df, y="marks")
plt.title("Marks Box Plot")
plt.show()
````

### Violin Plot — distribution + density

### Violin Plot
**What is it?** A violin plot combines information about the spread of data with its estimated density.

**What does this plot show?** The distribution and density of students' marks.

**How does it help us?** The shape shows where observations are more concentrated, giving more distribution detail than a basic box plot.

**Remember:** Use `violinplot()` when you want to understand the shape of a distribution.

````python
sns.violinplot(data=df, y="marks")
plt.title("Marks Violin Plot")
plt.show()
````

## 4. Basic Customization

### Customizing a Plot
**What is this cell doing?** We add `hue="department"` so the points are grouped by department, and we also customize the figure size, title, and axis labels.

**Why does this help?** Customization makes a plot easier to read and allows us to compare groups within the same visualization.

**Remember:** `hue` is very useful when you want to add a categorical grouping to a plot.

````python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="hours",
    y="marks",
    hue="department"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
````

## 5. Statistical Plots

### Pair Plot

### Pair Plot
**What is it?** A pair plot creates multiple plots to show relationships between several numerical variables.

**What does this plot show?** Relationships among sales, study hours, and marks.

**How does it help us?** It gives us a quick overview of many variable-to-variable relationships in one visualization.

**Remember:** Use `pairplot()` for quick exploratory data analysis when you have several numerical columns.

````python
sns.pairplot(df[["sales", "hours", "marks"]])
plt.show()
````

### Correlation Heatmap

### Correlation Heatmap
**What is it?** A heatmap represents numerical values using shades and annotations.

**What does this plot show?** The correlation between sales, study hours, and marks.

**How does it help us?** It makes relationships between multiple variables easy to compare at a glance. Values closer to `1` indicate a strong positive relationship, values closer to `-1` indicate a strong negative relationship, and values near `0` indicate little linear relationship.

**Remember:** A heatmap is especially useful for correlation matrices and other matrix-shaped data.

````python
corr = df[["sales", "hours", "marks"]].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
````

### Regression Plot

### Regression Plot
**What is it?** A regression plot shows a scatter plot together with a fitted regression line.

**What does this plot show?** The relationship between study hours and marks, along with a trend line.

**How does it help us?** It helps us see the overall direction of the relationship and whether the points roughly follow a linear trend.

**Remember:** Use `regplot()` when you want to visualize a relationship together with a simple regression trend.

````python
sns.regplot(data=df, x="hours", y="marks")
plt.title("Study Hours vs Marks")
plt.show()
````

## 6. Categorical Plots

### Count Plot

### Count Plot
**What is it?** A count plot counts how many observations belong to each category.

**What does this plot show?** The number of records in each department.

**How does it help us?** It quickly tells us how many observations we have in each category.

**Remember:** Use `countplot()` when you want category frequencies.

````python
sns.countplot(data=df, x="department")
plt.title("Department Count")
plt.show()
````

### Strip Plot

### Strip Plot
**What is it?** A strip plot displays individual observations as points across categories.

**What does this plot show?** Individual marks for each department.

**How does it help us?** Instead of only seeing an average or summary, we can see the actual observations and how they are distributed within each category.

**Remember:** Use `stripplot()` when individual data points matter.

````python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [100, 120, 150, 140, 170, 200],
    "hours": [2, 3, 4, 5, 6, 7],
    "marks": [40, 40, 40, 25, 42, 85],
    "department": ["IT", "IT", "IT", "Sales", "HR", "IT"]
})

df
````

````python
sns.stripplot(data=df, x="department", y="marks")
plt.title("Marks by Department")
plt.show()
````

### Swarm Plot

### Swarm Plot
**What is it?** A swarm plot also displays individual observations, but positions the points to reduce overlap.

**What does this plot show?** Individual marks for each department with the points arranged more clearly.

**How does it help us?** It makes individual observations easier to see when several points have similar values.

**Remember:** Use `swarmplot()` when you want to show individual observations while reducing point overlap.

````python
sns.swarmplot(data=df, x="department", y="marks")
plt.title("Marks by Department")
plt.show()
````

## 7. Saving a Plot

### Scatter Plot
**What is it?** A scatter plot places one observation as a point using an x-value and a y-value.

**What does this plot show?** The relationship between study hours and marks.

**How does it help us?** It helps us see whether two numerical variables have a relationship, such as whether marks tend to increase as study hours increase.

**Remember:** Use `scatterplot()` to investigate relationships between two numerical variables.

````python
sns.scatterplot(data=df, x="hours", y="marks")
plt.title("Study Hours vs Marks")

plt.savefig("study_hours_vs_marks.png", dpi=300)
plt.show()
````

## 8. Quick Cheat Sheet

| Plot | Use |
|---|---|
| `lineplot()` | Trends |
| `barplot()` | Category comparison |
| `histplot()` | Distribution |
| `kdeplot()` | Smooth distribution |
| `scatterplot()` | Relationship |
| `boxplot()` | Spread / outliers |
| `violinplot()` | Distribution + density |
| `pairplot()` | Relationships between many variables |
| `heatmap()` | Correlation / matrix values |
| `regplot()` | Relationship + regression line |
| `countplot()` | Category counts |
| `stripplot()` | Individual categorical points |
| `swarmplot()` | Individual points with less overlap |

### Most important Seaborn pattern

```python
sns.scatterplot(
    data=df,
    x="column1",
    y="column2",
    hue="column3"
)
```

`data` = DataFrame  
`x` = X-axis column  
`y` = Y-axis column  
`hue` = split/group using color
